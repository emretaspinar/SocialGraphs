```python
import pandas as pd, networkx as nx

nodes = pd.read_csv("../network/films_nodes.tsv", sep="\t", comment="#",
                    quoting=3)  # QUOTE_NONE: TSVs never quote, but plots may contain "quotes"
edges = pd.read_csv("../network/films_edges.tsv", sep="\t", comment="#", quoting=3)

G = nx.Graph()  # undirected: a shared director/actor links both films
G.add_nodes_from(nodes.node_id)
G.add_weighted_edges_from(edges[["source", "target", "weight"]].itertuples(index=False))
# G: 1789 nodes, 11802 edges
```
